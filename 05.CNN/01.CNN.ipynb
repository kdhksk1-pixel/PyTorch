```python
import torch
from torch import nn
```

```python
layer=nn.Conv2d(in_channels=1, out_channels=2, kernel_size=3)   # stride=1, padding=0 디폴트
layer(torch.randn(32, 1, 5, 5)).shape
```

```python
layer=nn.Conv2d(3, 5, 3)
print(layer(torch.randn(32, 3, 5, 5)).shape)

print(layer.weight.shape)       # [특징(출력), 채널(입력), 행, 열]
```

```python
conv1=nn.Conv2d(1, 8, 6, stride=2)
x=torch.randn(32, 1, 28, 28)
print(conv1(x).shape)

conv2=nn.Conv2d(8, 16, 3, padding=1)
print(conv2(conv1(x)).shape)

maxpool=nn.MaxPool2d(kernel_size=2, stride=2)
print(maxpool(conv2(conv1(x))).shape)
```

```python
maxpool=nn.MaxPool2d(2)     # kernel_size=2, stride=2와 동일
x=torch.rand(1, 6, 6)
print(x)
print(maxpool(x))
print(maxpool(torch.randn(32, 3, 6, 6)).shape)
```

```python
maxpool=nn.AvgPool2d(2)
x=torch.randn(1, 6, 6)
print(x)
print(maxpool(x))
print(maxpool(torch.randn(32, 3, 6, 6)).shape)
```

```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1=nn.Conv2d(1, 8, 6, stride=2)
        self.conv2=nn.Conv2d(8, 16, 3, padding=1)
        self.Maxpool=nn.MaxPool2d(2)
        self.fc=nn.Linear(16*6*6, 10)

    def forward(self, x):
        x=self.conv1(x)
        print('conv1:', x.shape)
        x=self.conv2(x)
        x=self.Maxpool(x)
        x=torch.flatten(x, start_dim=1)
        x=self.fc(x)
        return x

x=torch.randn(32, 1, 28, 28)
model=CNN()
print(model(x).shape)
```
