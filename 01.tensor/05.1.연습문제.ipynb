#### 문제 02

실험 장비에서 측정한 양수 데이터가 다음과 같음.

```python

```

1. 각 원소의 제곱근을 `sqrt_data`에 저장.
2. 각 원소의 제곱을 `square_data`에 저장.

1. 전체 생산량의 합계와 평균 계산.
2. 팀별 생산량 합계를 `dim=1` 기준으로 계산.
3. 팀별 평균 생산량을 `dim=1` 기준으로 계산하고 `keepdims=True` 적용.

**요구사항**
1. 각도가 30°, 45°, 60°인 값을 라디안으로 변환한 텐서 생성.
2. 각도 텐서의 사인과 코사인 계산.
3. 45°의 탄젠트 계산.
4. `[-1, 0, 1]`을 입력으로 아크사인, 아크코사인, 아크탄젠트 계산.
5. `[-3., -1., 0., 1., 3.]`에 `torch.tanh()` 적용.

````python
angles=torch.tensor([30., 45., 60.])
rad_angles=torch.deg2rad(angles)

sin_val=torch.sin(rad_angles)
cos_val=torch.cos(rad_angles)

tan_45=torch.tan(torch.deg2rad(torch.tensor(45.)))

inputs=torch.tensor([-1., 0., 1.])
asin_val=torch.asin(inputs)
acos_val=torch.acos(inputs)
atan_val=torch.atan(inputs)

tanh_inputs=torch.tensor([-3., -1., 0., 1., 3.])
tanh_val=torch.tanh(tanh_inputs)

print("라디안 변환 텐서:", rad_angles)
print("사인 값:", sin_val)
print("코사인 값:", cos_val)
print("45도 탄젠트 값:", tan_45.item())
print("아크사인", asin_val)
print("아크코사인", acos_val)
print("아크탄젠트", atan_val)
print("하이퍼볼릭 탄젠트:", tanh_val)
````

```python

```

**요구사항**
1. `transpose()`를 이용해 0번 차원과 2번 차원의 위치를 교환.
2. `permute()`를 이용해 차원 순서를 `(1, 2, 0)`으로 재배치.
3. 두 결과의 Shape 출력.
4. 별도의 2차원 행렬을 생성한 뒤 `.t()`와 `.T`를 이용해 전치 결과 출력.

```python
sensor = torch.arange(6).reshape(1, 3, 1, 2, 1)
```

2. `squeeze()`를 이용해 크기가 1인 모든 차원 제거.
3. `squeeze(dim=0)`과 `squeeze(dim=2)`의 결과 Shape을 각각 확인.

#### 문제 13

두 학급의 시험 점수를 나타내는 텐서가 다음과 같음.

```python
class_a = torch.tensor([[70, 80, 90],
                        [65, 75, 85]])
class_b = torch.tensor([[60, 70, 80],
                        [55, 65, 75]])
```

**요구사항**
1. `torch.vstack()`으로 두 텐서를 세로 방향으로 결합.
2. `torch.hstack()`으로 두 텐서를 가로 방향으로 결합.
3. `torch.cat()`의 `dim=0`, `dim=1` 결과 출력.
4. `torch.stack()`으로 새로운 차원을 추가하여 결합.
5. 각 결과의 Shape 출력.

````python
class_a = torch.tensor([[70, 80, 90],
                        [65, 75, 85]])
class_b = torch.tensor([[60, 70, 80],
                        [55, 65, 75]])

v_stacked = torch.vstack((class_a, class_b))

h_stacked = torch.hstack((class_a, class_b))

cat_dim0 = torch.cat((class_a, class_b), dim=0)
cat_dim1 = torch.cat((class_a, class_b), dim=1)

stacked = torch.stack((class_a, class_b))

print("원본 Shape:", class_a.shape) # (2, 3)
print("-" * 30)
print("1. vstack Shape:", v_stacked.shape)       # (4, 3)
print("2. hstack Shape:", h_stacked.shape)       # (2, 6)
print("3-1. cat(dim=0) Shape:", cat_dim0.shape)  # (4, 3)
print("3-2. cat(dim=1) Shape:", cat_dim1.shape)  # (2, 6)
print("4. stack Shape:", stacked.shape)          # (2, 2, 3)
````

PyTorch 텐서를 NumPy 배열로 변환하고, 자동 미분 그래프와 연결된 텐서도 NumPy로 변환하려고 함.

**요구사항**
1. CPU 텐서 `x`를 생성하고 `.numpy()`로 NumPy 배열 변환.
2. 변환한 NumPy 배열의 자료형과 Shape 출력.
3. NumPy 배열의 원소를 변경하고 원본 텐서의 값도 변경되는지 확인.
4. `requires_grad=True`인 텐서 `y` 생성.
5. `y`에 대한 연산을 수행하고 `backward()` 실행.
6. `y.detach().cpu().numpy()`를 이용해 NumPy 배열로 변환.

````python
x = torch.tensor([1.0, 2.0, 3.0])
x_np = x.numpy()

print("x_np 자료형:", x_np.dtype)
print("x_np shape:", x_np.shape)

x_np[0] = 99.0
print("\n변경된 x_np:", x_np)
print("원본 텐서 x:", x) 

y = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)

z = y.sum()
z.backward()
print("\ny의 기울기(grad):", y.grad)

y_np = y.detach().cpu().numpy()
print("변환된 y_np:", y_np)
````

#### 문제 18

학습 시간과 문제 풀이 횟수로 시험 점수를 예측하기 위한 데이터를 준비하려고 함.

**요구사항**
1. `pandas.DataFrame`으로 8명의 학습 시간, 문제 풀이 횟수, 시험 점수 데이터를 생성.
2. 시험 점수를 제외한 입력 데이터 `x`와 시험 점수 `y`를 분리.
3. `.values`를 이용해 NumPy 배열로 변환.
4. `y`를 `reshape(-1, 1)`로 2차원 구조로 변경.
5. 입력과 정답을 `torch.float32` 텐서로 변환.
6. `TensorDataset`을 생성하고 데이터 건수 및 첫 3개 데이터 확인.
7. `DataLoader`를 `batch_size=3`, `shuffle=False`로 생성하고 배치 수와 각 배치 데이터를 출력.
8. `shuffle=True`인 DataLoader도 생성하여 첫 번째 배치 확인.

````python
import pandas as pd
import torch
from torch.utils.data import TensorDataset, DataLoader

data = {
    '학습시간': [2, 4, 6, 8, 10],
    '문제풀이횟수': [10, 20, 30, 40, 50],
    '시험점수': [50, 60, 70, 80, 90]
}
df = pd.DataFrame(data)


X_df = df[['학습시간', '문제풀이횟수']]
y_df = df['시험점수']

X_np = X_df.values
y_np = y_df.values

y_np = y_np.reshape(-1, 1)

X = torch.tensor(X_np, dtype=torch.float32)
y = torch.tensor(y_np, dtype=torch.float32)

dataset = TensorDataset(X, y)
print("총 데이터 건수:", len(dataset))
print("첫 3개 데이터:\n", dataset[:3])

dataloader = DataLoader(dataset, batch_size=2, shuffle=False)
print(f"\n총 배치 수: {len(dataloader)}")
for i, batch in enumerate(dataloader):
    print(f"배치 {i+1}:", batch)

shuffled_dataloader = DataLoader(dataset, batch_size=2, shuffle=True)
first_batch = next(iter(shuffled_dataloader))
print("\n셔플된 첫 번째 배치:\n", first_batch)
````
