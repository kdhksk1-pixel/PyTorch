#### 문제 01

`nn.Linear()`를 사용하여 입력 특징 3개를 출력 특징 2개로 변환하기.

**조건**
- 입력 데이터 x:
  `[[1., 2., 3.], [2., 0., 1.]]`
- Weight:
  `[[1., 0., -1.], [2., 1., 0.]]`
- Bias: `[0.5, -1.]`

**요구사항**
1. `nn.Linear(3, 2)`로 선형 계층 생성.
2. 계층의 Weight와 Bias를 조건에 맞게 설정.
3. 입력 데이터 x를 모델에 전달하여 출력 y 계산.
4. Weight, Bias, 입력 및 출력의 Shape 출력.
5. 최종 출력값 확인하기.

```python
import torch
import torch.nn as nn
```

```python
model = nn.Linear(3, 2)
model.weight = nn.Parameter(torch.tensor([[1., 2., 3.], [2., 0., 1.]]))
model.bias = nn.Parameter(torch.tensor([0.5, -1.]))

x = torch.tensor([[1., 2., 3.], [2., 0., 1.]])

y = model(x)

print('weight :', model.weight)
print('w shape:', model.weight.shape)
print('bias :', model.bias)
print('bias shape :', model.bias.shape)
print('output :', y)
print('output shape :', y.shape)
```

#### 문제 02

`nn.Sequential()`을 이용하여 여러 선형 계층을 순서대로 연결하기.

**모델 구조**
1. 입력 특징 4개 → 출력 특징 6개
2. Sigmoid 활성화 함수
3. 입력 특징 6개 → 출력 특징 3개
4. Sigmoid 활성화 함수
5. 입력 특징 3개 → 출력 특징 2개

**요구사항**
1. 위 구조대로 `nn.Sequential()` 모델 생성.
2. 데이터 5개를 포함하고 각 데이터의 특징이 4개인 입력 텐서 x 생성.
3. 모델에 x를 전달하여 출력 y 계산.
4. 입력과 출력의 Shape 출력.
5. 각 Linear 계층의 Weight와 Bias Shape 출력.
6. 모델 전체의 학습 가능한 파라미터 개수 출력하기.

```python
model = nn.Sequential(nn.Linear(4, 6),
                      nn.Sigmoid(),
                      nn.Linear(6, 3),
                      nn.Sigmoid(),
                      nn.Linear(3, 2)
                      )

x = torch.randn(5, 4)
y = model(x)

print('input shape :', x.shape)
print('output shape :', y.shape)

# model.named_moudules() : 모델안에 있는 모든계층의 이름과 계층 객체를 가져오는 함수
for name, layer  in model.named_modules():
    print(name, ':', layer)
    # if isinstance(layer, nn.Linear) :
    #     print('weight shape :', layer.weight.shape)
    #     print('bias shape', layer.bias.shape)

num_param = sum(
    p.numel() for p in model.parameters()
)
print(num_param)
```
