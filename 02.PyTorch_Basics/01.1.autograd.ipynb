#### ▶ Autograd(자동미분)
 : PyTorch에서 연산과정을 추적하고 역전파를 통해 미분값을 자동으로 계산하는 기능
 - **미분 계산 자동화**
 - **기울기 계산**: 각 입력값 모델의 가중치가 손실함수에 미치는 영향 계산
 - **모델 학습 지원**: 계산한 기울기를 Optimizer에 전달하여 가중치 업데이트

````python
import torch
````

##### Autograd의 기본 흐름
1. 미분할 텐서에 requires_grad=True설정
2. 해당 텐서를 사용하여 연산 수행
3. 결과 텐서에 backward() 실행
4. 원본 텐서의 grad속성에서 미분값 확인

#### 1. requires_grad
- True: 해당 텐서에 대한 연산을 추적하여 기울기를 계산할 수 있도록 설정.
- False: 기본값, 해당 텐서에 대한 연산을 추적하지 않음
- 일반적으로 모데링 학습해야 하는 가중치(weight)와 편향(bias)에 True를 설정함.

````python
x=torch.tensor(1.0, requires_grad=True)
print(x)
````

````python
x=torch.tensor(1.0)
print(x)
print(x.requires_grad)

x.requires_grad=True
print(x)
print(x.requires_grad)
````

````python
# 1.
x=torch.tensor([1.0], requires_grad=True)

# 2. 함수 계산
y=x**2
print(y)

# 3. 역전파 실행
y.backward()

# 4. 결과와 기울기 확인
print('y:', y)
print('x의 기울기:', x.grad)
````

````python
x=torch.tensor(1.0, requires_grad=True)
x.requires_grad=False
    # transfer Leanrning할 때 사용
    # 만들어 놓은 학습된 모델ㄹ에서 일부분만 학습시키고 싶을 때 나머지 부분만
y=x**2
print(y)
# y.backward() # error
````

````python
x=torch.tensor([2.0, 4.0], requires_grad=True)
y=x*3

print('x.requires_grad:', x.requires_grad)
print('y.requires_grad:', y.requires_grad)
print('x.is_leaf:', x.is_leaf)
print('y.is_leaf:', y.is_leaf)
print('y.grad_fn:', y.grad_fn)
````

#### 2. 연쇄법칙(chain Rule)
함수가 여러 연산을 연결되어 있으면 각 연산의 미분값을 연쇄적으로 곱하여 최종 기울기를 계산함.

````python
x=torch.tensor(1.0, requires_grad=True)
y=x**2
print(y)

z=3*y
print(z)

z.backward()
print(x.grad)
print(y.grad)
````

````python
# 중간에 있는 y의 미분값도 알고 싶다면
x=torch.tensor(1., requires_grad=True)
y=x**2
y.retain_grad()

z=3*y

z.backward()
print('x의 미분값:', x.grad)
print('y의 미분값:', y.grad)
````

````python
x=torch.tensor(1., requires_grad=True)
y=x**2
z=3*y

y.backward()    # y부터 미분 시작(z는 안함.)
print(x.grad)
````

````python
x=torch.tensor(1., requires_grad=True)
a=x**2
b=a+1
print(b)

c=b**2
c.backward()
print(x.grad)
````

#### ▶ 3. 편미분
 : 함수에 여러 변수가 있을 때는 각 변수에 대해 따로 미분할 수 있음.

````python
x=torch.tensor(1., requires_grad=True)
y=torch.tensor(1., requires_grad=True)

z=2*x**2 + y**2
print(z)

z.backward()
print(x.grad)
print(y.grad)
````

````python
x=torch.tensor(2., requires_grad=True)
y=torch.tensor(3., requires_grad=True)

f=x**2 + 2*x*y + y**2
f.backward()

print(x.grad)
print(y.grad)
````

#### ▶ 4. 벡터 출력의 기울기 계산
- 방법 1.

y.sum().backward()

- 방법 2.

y.backward(torch.ones_like(y))

````python
x=torch.tensor([1., 2, 3], requires_grad=True)
y=torch.sum(x**2)

y.backward()
print('y:', y)
print('x.grad:', x.grad)
print('-'*50)

x=torch.tensor([1., 2, 3], requires_grad=True)
y=x**2
y.sum().backward()
print('y:', y)
print('x.grad:', x.grad)
print('-'*50)

x=torch.tensor([1., 2, 3], requires_grad=True)
y=x**2
y.backward(torch.ones_like(y))
````

#### 5. Gradient 누적과 초기화
 : backward()를 여러번 실행하면 grad에 새로운 기울기가 더해짐.

 - 기울기를 새로 계산하려면 기존 기울기를 0으로 초기화
    - x.grad.zero_()
    - 모델 학습에서는 일반적으로 optimizer.zero_grad()를 사용함.

````python
x=torch.tensor([1., 2, 3], requires_grad=True)

y=(x**2).sum()
y.backward()
print('첫 번째 backward:', x.grad)

z=(3*x).sum()
z.backward()
print('두 번째 backward 후 누적:', x.grad)

x.grad.zero_()
print('초기화 후:', x.grad)

w=(x**3).sum()
w.backward()
print('초기화 후 새로 계산:', x.grad)
````

#### 6. torch.autograd.grad()
 : 지정한 입력 텐서에 대한 기울기(미분값)를 반환하는 함수.

 - backward(): 계산한 기울기를 입력 텐서의 .grad에 누적
 - torch.autograd.grad(): 기울기를 반환해줌. 기본적으로 .grad에 누적하지 않음.

 - 언제 사용?
    1. 특정 변수의 미분값만 필요할 때
    2. 기울기를 .grad에 저장하고 싶지 않을 때
    3. 여러 변수를 편미분을 동시에 구할 때: 여러 입력 변수의 기울기를 튜플로 반환받을 수 있음.

- 사용 형식
torch.autograd.grad(outputs, inputs)

````python
x=torch.tensor(4., requires_grad=True)
y=x**3

gradient=torch.autograd.grad(y, x)

print('y:', y)
print('반환된 기울기:', gradient)
print('x.grad:', x.grad)
````

#### 7. detach()
requires_grad=True인 텐서에 연산을 적용하면 PyTorch는 연산과정을 추적함.<br>
그런데 특정 텐서부터는 더이상 Gradient를 계산하지 않도록 하고 싶을 때 detach()를 사용
**기존 텐서와 값은 공유하지만, 자동 미분 계산 그래프에서 분리된 텐서를 반환**

- 원본 텐서의 연산 추적은 그대로 유지됨
- 분리된 텐서는 해당 연산 그래프를 통한 기울기 계산에 연결되지 않음
- 학습 결과를 출력하거나 Numpy로 변환할 때 자주 사용함.

````python
x=torch.tensor([1., 2, 3], requires_grad=True)
y=x*2
y_detach=y.detach()

print('y:', y)
print('y_requires_grad:', y.requires_grad)
print('y_detach:', y.detach)
print('y_detach.requires_grad:', y_detach.requires_grad)
print('y_detach.grad_fn:', y_detach.grad_fn)
````

````python
# 변수의 값을 그냥 상수로 사용하고 싶을 때
x=torch.tensor(1., requires_grad=True)
y=x**2
z=y.detach()        # y값을 상수로 넣음

w=y+z    # x**2 + 1 이렇게 들어감.
w.backward()
print(x.grad)
````

#### 8. torch.no_grad()
 : 지정한 블록 코드 안에서는 자동 미분을 위한 연산 추적을 비활성화함.

 - 학습 중 모델의 가중치를 업데이트할 때
 - 검증 및 예측처럼 기울기 계산이 필요하지 않을 때
 - 불필요한 계산 그래프 생성을 줄여 메모리 사용량을 줄일 수 있음.

 [사용 형식]
 ```python
 with torch.no_grad():
    # 기울기를 추적하지 않은 연산
```

- `detach()`는 특정 텐서를 계산 그래프에서 분리하고, `no_grad`는 특정 코드 블록의 연산 추적을 막는다는 차이가 있음.

````python
x=torch.tensor(3., requires_grad=True)
y1=x**2

with torch.no_grad():
    y2=x**2
    z=2*y2

print('일반 연산 requires_grad:', y1.requires_grad)
print('no_grad 내부 연산 y2의 requires_grad:', y2.requires_grad)
print('no_grad 내부 연산 z의 requires_grad:', z.requires_grad)
print('y2.grad_fn:', y2.grad_fn)
````

- torch.no_grad()
    - 특정 코드 블록에 적용
    - 블록 내부의 연산 그래프 추적을 중단
    - 블록이 끝나면 기존 상태로 복귀
- requires_grad=False
    - 특정 텐서에 적용
    - 해당 텐서에 대한 기울기 추적을 비활성화
    - 다시 True로 변경하기 전까지 유지.

````python
x=torch.tensor(1., requires_grad=True)

with torch.no_grad():
    y=x**2
    print(x.requires_grad)
    print(y)
print(x.requires_grad)

y=x**2
print(y)
print(y.requires_grad)
print('-'*30)

x=torch.tensor(1., requires_grad=True)
x.requires_grad=False
y=x**2
print(x.requires_grad)
print(y)
````
