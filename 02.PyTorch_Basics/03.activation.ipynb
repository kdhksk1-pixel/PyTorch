### Activation Function
: 신경망의 선형 연산 결과에 비선형성을 추가하여 복잡한 패턴을 학습할 수 있도록 돕는 함수

- 활성화 함수의 목적
    - 은닉층에 비선형성 추가
    - 출력값을 조절
    - 문제에 맞는 출력 생성

#### 1. Sigmoid 함수
: 입력값 0~1사이의 값으로 변환하는 함수

- 특징
    - 출력 범위가 0과 1 사이
    - 입력값이 0이면 출력값은 0.5
    - 입력값이 커지면 출력은 1에 가까워짐
    - 입력값이 작아지면 출력은 0에 가까워짐
    - 이진분류에서 한 클래스에 속할 확률처럼 해석할 수 있는 값들을 만들때 사용

- 단점
    - 입력값의 절대값이 아주 커지면 기울기 작아지는 포화현상이 나타남
    - 은닉층에 여러번 사용하면 기울기 소실 문제가 생길수 있음

```python
import torch
import torch.nn as nn
```

```python
x = torch.tensor([-10., -2, -1, 0, 1, 2, 10])

output = torch.sigmoid(x)

print(output)
```

```python
logits = torch.tensor([-2, -0.5, 0, 0.8, 2])
print('모델을 통과한 출력(Logits) :', logits)

si = torch.sigmoid(logits)
print('Sigmoid 출력 :', si)

class0 = si >= 0.5
print('분류결과 :', class0)

class0 = (si >= 0.5).int()
# .int() Boolean값을 정수형으로 변환
#       False -> 0
#       True -> 1
print('분류결과 :', class0)
```

#### 2. ReLu 함수
: 입력값이 0보다 작으면 0을 출력하고, 0보다 크면 입력값을 그대로 출력

- 특징
    - 계산이 단순하고 효율적임
    - 양수 영역에서는 기울기가 1이므로 기울기 소실 문제 완화됨
    - 일반적인 신경망의 은닉층에서 널리 사용됨

- 단점
    - 음수 영역에서는 출력과 기울기가 0임
    - 학습 중 일부 뉴런이 음수 영역에 머물면 업데이트가 잘 되지 않음

```python
# nn.ReLU()
x = torch.tensor([-5., -2, -1, 0, 1, 2, 5])
output = torch.relu(x)

print('입력 : ',x)
print('ReLU 출력 :', output)
```

#### 3. Tanh
: 입력값이 -1 ~ 1사이의 값으로 변환

- 특징
    - 입력값이 0이면 출력값도 0
    - 양수입력은 1에 가깝고, 음수 입력은 -1에 가까움
    - 출력중심이 0
- 단점
    - 입력값의 절대값이 매우 커지면 기울기가 작아지는 포화현상이 발생

```python
x =torch.tensor([-10., -2, -1, 0, 1, 2, 10])

output = torch.tanh(x)

print('입력 :', x)
print('Tanh 출력 :', output)
```

#### 4. Softmax
: 입력받은 실수 값들으 0과 1사이의 값으로 변환하고, 이 값들의 총합이 정확히 1이 되도록 만들어 확률분포로 해석할 수 있게 해줌<br>
주로 서로 배타적인 여러 클래스 중 하나를 선택하는 다중분류의 출력에 사용됨

- 특징
    - 각 출력값든 0~1사이의 값
    - 모든 클래스를 출력값을 합치면 1이됨
    - 출력값이 각 클래스에 대한 예측 확률처럼 해석할 수 있음
    - 가장 큰 출력값을 가진 클래스가 모델의 예측 클래스가 됨

- 옵션 dim
    - dim=0 : 첫번째 차원을 기준으로 Softmax계산
    - dim=1 : 두번째 차원을 기준으로 Softmax계산
    - Batch데이터가 (batch_size, num_classes) 형태라면 일반적으로 dim=1을 사용함

```python
logits = torch.tensor([2, 1, 0.1])

sm = torch.softmax(logits, dim=0)

print('Logits :', logits)
print('Softmax :', sm)
print('확률의 합 :', sm.sum())

# 가장 높은 확률을 가진 클래스를 선택
print('예측 클래스 인덱스 :', torch.argmax(sm))
print(torch.argmax(logits))
```

- 예를 들어 데이터 2개를 각각 3개 클래스로 분류한다면 shape(2, 3)
    - 각 행(row)은 하나의 데이터에 대한 클래스의 합이 1이 되어야 함
    - 각 데이터별로 클래스 확률의 합이 1이 되어야 함

- 비교
    - dim=0 : 같은 클래스 열을 기준으로 서로 다른 데이터 사이의 합이 1이됨
    - dim=1 : 각 데이터(행)의 클래스 확률 합이 1이 됨

```python
# 2개의 데이터(사람)가 있고, 각 3개의 클래스에 대한 예측 점수를 가지고 있다
logits = torch.tensor([
    [2.0, 1.0, 0.1],
    [0.5, 2.5, 1.0]
])

prob_dim0 = torch.softmax(logits, dim=0)
prob_dim1 = torch.softmax(logits, dim=1)

print('dim=0의 결과')
print(prob_dim0)
print('dim=0 열 방향의 합 :', prob_dim0.sum(dim=0))
print('-'* 30)

print('dim=1의 결과')
print(prob_dim1)
print('dim=1 행 방향의 합 :', prob_dim1.sum(dim=1))
```

#### 4가지 활성화 함수 비교
| 함수 | 출력범위 | 대표적인 사용 위치 |
| --- | --- | --- |
| Sigmoid | 0 ~ 1 | 이진분류 출력층 |
| ReLU | 0 ~ 양의 무한대 | 일반적인 은닉층 |
| Tanh | -1 ~ 1 | 일부 은닉층 |
| Softmax | 0~1 전체합=1 | 다중 분류 출력층 |

```python
x = torch.tensor([-2., -1, 0, 1, 2])

print('Sigmoid :', torch.sigmoid(x))
print('ReLU :', torch.relu(x))
print('Tanh :', torch.tanh(x))

# Softmax는 여러 클래스의 점수에 적용
class_scores = torch.tensor([2., 1, 0.1])
print('Softmax :', torch.softmax(class_scores, dim=0))
```
