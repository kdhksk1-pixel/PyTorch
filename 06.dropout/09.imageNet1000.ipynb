```python
import torch
from torch import nn
import torch.nn.functional as F
from torchvision import models, transforms
import matplotlib.pyplot as plt
from torchinfo import summary
```

```python
weights=models.ResNet50_Weights.IMAGENET1K_V2
load_model=models.resnet50(weights=weights)
print(load_model)
```

```python
load_model.eval()
with torch.no_grad():
    print(load_model(torch.randn(2, 3, 224, 224)).shape)
```

```python
# 모델의 전체 구조, 각 층의 출력 크기, 파라미터 개수, 연산량 등을 확인하는 함수
summary(load_model, input_size=(2, 3, 224, 224))
```

```python
transform=transforms.Compose([
    transforms.ToTensor(),
    transforms.Resize(size=(256)),
    transforms.CenterCrop(size=(224, 224))
])
```

```python
# meta는 사전학습된 가중치에 대한 부가정보를 딕셔너리 형태로 저장하고 있다
print(weights.meta.keys())
```

```python
idx2label=dict(enumerate(weights.meta['categories']))
print(idx2label)
```

```python
import imageio.v2 as imageio

X=imageio.imread('https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcS-UVQylXApHLhSkyPNzVCHXCN6KQdBve4tvl9hpC2EHQ&s=10')
print(type(X))
print(X.shape)
plt.imshow(X)

X=transform(X)
print(type(X))
print(X.shape)
plt.figure()
plt.imshow(X.permute(1, 2, 0))

X=X.unsqueeze(dim=0)
print(X.shape)
```

```python
load_model.eval()
with torch.no_grad():
    y_hat_sorted=load_model(X).sort(descending=True)
    percent_top5=F.softmax(y_hat_sorted[0], dim=1)[0, :5]*100
    print(percent_top5)

    pred_top5=y_hat_sorted[1][0, :5]
    print(pred_top5)
for i, idx in enumerate(pred_top5):
    print(f'{idx2label[idx.item()]} ({round(percent_top5[i].item(), 1)}%)')
```

```python
load_model.eval()
features=nn.Sequential(*list(load_model.children())[:-2])

with torch.no_grad():
    X_features=features(X)

plt.figure(figsize=(10, 5))
for i in range(6):
    plt.subplot(2, 3, i+1, xticks=[], yticks=[])
    plt.imshow(X_features[0, i])
    plt.colorbar()

summed_map=X_features.sum(dim=1)

plt.figure(figsize=(8, 8))
plt.xticks([])
plt.yticks([])
plt.imshow(X[0].permute(1, 2, 0))
plt.imshow(summed_map[0], extent=[0, 224, 224, 0], alpha=0.4)

pred_class=idx2label[pred_top5[0].item()]
plt.title(f'{pred_class} ({round(percent_top5[0].item(), 1)}%)')
```
